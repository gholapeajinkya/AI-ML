# Transformers, what can they do?

[Transformer](https://huggingface.co/learn/llm-course/chapter1/3) models are used to solve all kinds of tasks across different modalities, including natural language processing (NLP), computer vision, audio processing, and more.

The most basic object in the 🤗 Transformers library is the pipeline() function. It connects a model with its necessary preprocessing and postprocessing steps, allowing us to directly input any text and get an intelligible answer:

```python
from transformers import pipeline
```

```python
classifier = pipeline("sentiment-analysis")
classifier("I love using transformers for natural language processing tasks!")
```

```python
classifier(
    [
        "I've been waiting for a HuggingFace course my whole life.", 
        "I hate this so much!"
    ]
)
```

## Available pipelines for different modalities
The pipeline() function supports multiple modalities, allowing you to work with text, images, audio, and even multimodal tasks. 

### Text pipelines
- text-generation: Generate text from a prompt
- text-classification: Classify text into predefined categories
- Summarization: Create a shorter version of a text while preserving the key information
- translation: Translate text from one language to another
- zero-shot-classification: Classify text without prior training on specific labels

### Imape pipelines
- image-to-text: Generate text description of image
- imgae-classification: Identify object in an image
- object-detection: Locate and identify object in image

### Audio pipeline
- automatic-speech-recognition: Convert speech to text
- audio-classification: Classify audio into categories
- text-to-speech: Convert text to spoken audio

### Multimodal pipeline
- image-text-to-text: Respond to an image based on text prompt

## Zero-shot-classification

We'll start by tackling a more challenging task where we need to classify texts that haven't been labelled. This is a common scenario in real-world projecs because annotating text is usually time-consuming and requires domain experties. For this use case, the 'zero-shot-classification' pipeline is very powerful: it allows you to specify which labels to use for the classification, so you don't have to rely on the labels of pretarined model.

We have already seen how the model can classify a sentence as positive or negative using those two labels -  but it can also classify the text using any set of labels you line

```python
classifier = pipeline("zero-shot-classification")
classifier(
    "This is a course about the Transformers library",
    candidate_labels=["education", "politics", "business"]
)
```

## Mask filling

The next pipeline we will try is 'fill-mask'. The idea of this task is to fill in the balnks in a given text:

```python
unmask = pipeline("fill-mask")
unmask("Transformers is a <mask> library for natural language processing.", top_k=2)
```
