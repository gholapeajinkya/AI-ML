# Linear Regression
Linear Regression is a fundamental supervise learning algorithm used to model the relationship between a dependent variable and one or mode independent variables. It predicts continuous values by fitting a straight line that best represents the data

## Types of Linear Regression
- Simple Linear Regression
- Multiple Linear Regression

## Linear Regression Example

Suppose you want to know if money makes people happy, so you download Better Life Index data from OECD's website & World Bank stats about gross domestic product (GDP) per capita

```python
import pandas as pd

# Download the data
data_root = "https://github.com/ageron/data/raw/main"
lifesat = pd.read_csv(f"{data_root}/lifesat/lifesat.csv")
lifesat.head()
```

```python
import matplotlib.pyplot as plt

# Visualize the data
X = lifesat[["GDP per capita (USD)"]].values
y = lifesat[["Life satisfaction"]].values

plt.scatter(X, y)
plt.xlabel("GDP per capita (USD)")
plt.ylabel("Life satisfaction")
plt.title("Life Satisfaction vs GDP per capita")
plt.show()
```

```python
# Select and train linear model
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)

# Make predictions
X_new = [[37_655.2]]  # Example GDP per capita value
y_pred = model.predict(X_new)
print(y_pred)
```

```python
# Plot the regression line
plt.scatter(X, y)
plt.plot(X, model.predict(X), color='red', linestyle="--")
plt.xlabel("GDP per capita (USD)")
plt.ylabel("Life satisfaction")
plt.title("Life Satisfaction vs GDP per capita with Regression Line")
plt.show()
```
